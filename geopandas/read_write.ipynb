```python
import pandas as pd
import geopandas as gpd
```

```python
# Create a dataframe with some sample data
data = {
    'Name': ['New York City', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix'],
    'Population': [8419600, 3980400, 2716000, 2328000, 1690000],
    'Latitude': [40.7128, 34.0522, 41.8781, 29.7604, 33.4484],
    'Longitude': [-74.0060, -118.2437, -87.6298, -95.3692, -112.0740]
}

data
```

```python
type(data)
```

```python
cities_df = pd.DataFrame(data)
cities_df
```

```python
gdf = gpd.GeoDataFrame(
  cities_df, geometry=gpd.points_from_xy(cities_df.Longitude, cities_df.Latitude))

gdf
```

```python
type(gdf)
```
