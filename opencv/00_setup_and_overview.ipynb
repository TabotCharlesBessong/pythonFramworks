## The one mental shift you need before anything else: vector vs. raster

Everything in the GeoPandas curriculum was **vector data** points, lines, polygons defined by exact
coordinates (a parcel boundary is a precise list of corner points). Everything in this curriculum is
**raster data** a grid of pixels, each with a color or intensity value, with no inherent "shapes" until
you compute them. A scanned title document and a satellite image are both just grids of numbers.

This distinction is *why* this is a genuinely different library stack, not just "more geospatial tools."

## The layers, same style as before

| Layer | Library | What it actually does | Where you'll use it |
|---|---|---|---|
| Basic image I/O | **Pillow (PIL)** | Opening, saving, and doing simple operations on ordinary image files (JPEG, PNG) | Loading a scanned document, reading EXIF metadata |
| General image processing | **OpenCV** | The workhorse filtering, edge detection, contour detection, feature matching, color space conversion | Most of document forensics: detecting tampering, deskewing scans, comparing image regions |
| Higher-level image algorithms | **scikit-image** | Similar territory to OpenCV, but with some particularly convenient, well-documented implementations of specific algorithms | Structural similarity (SSIM) for change detection one specific, very useful function |
| Geospatial raster I/O | **Rasterio** | Reads/writes *georeferenced* raster data satellite imagery that has real-world coordinates attached to each pixel, not just an ordinary photo | Loading actual satellite imagery, and critically the bridge back to your GeoPandas/Shapely parcel polygons, since you'll clip imagery to a parcel boundary |
| Deep learning | **PyTorch** | Building and training neural networks tensors, automatic differentiation, training loops | The eventual document-authenticity classifier and land-use segmentation, once real training data exists |
| Pretrained models & transforms | **Torchvision** | Ships pretrained models (trained on millions of general images) and the image transforms needed to use them | **Transfer learning** fine-tuning an existing model on your small dataset, instead of training from zero, which is unrealistic with limited real fraud/imagery examples |

## Mapping to the product's actual features

| Feature | What you'll build | Libraries |
|---|---|---|
| Document forensics (#1) | Error-level analysis, noise/copy-move detection, metadata checks rule-based checks that work with almost no training data, plus the classifier architecture for later once labeled examples accumulate | OpenCV, Pillow, PyTorch/torchvision |
| Satellite imagery verification (#3) | Reading and visualizing real georeferenced imagery, clipping it to a parcel boundary from your GeoPandas curriculum | Rasterio (+ GeoPandas/Shapely from before) |
| Land monitoring / change detection (#13) | Comparing two images of the same area over time to flag encroachment or unauthorized construction | scikit-image (SSIM), OpenCV |

## An honest note on training data, before you start

Notebook 04 (transfer learning) and any real trained classifier depend entirely on having actual labeled
examples genuine vs. forged documents, or before/after satellite imagery pairs. **This curriculum uses
synthetic, programmatically generated images throughout**, since real fraud examples and licensed
satellite imagery aren't available to build this with. The code you'll learn is the real code you'd run
— only the data is a stand-in, exactly as flagged in the product's implementation plan (the CV features
are correctly sequenced last in the roadmap for this exact reason).
